General function

```python
def make_ligand_count_dict(metal_tsv_string):
    metal_df = pd.read_csv(metal_tsv_string, sep = '\t')
    
    all_ligand_list = []
    for ligands in metal_df['Ligands']:
        temp = ligands.split(' ')
        all_ligand_list.append(temp)
    
    removed_extra_list = []
    for i in all_ligand_list:
        new_list = []
        for j in i:
            new_list.extend(re.findall(r'([A-Z]+)\_.*', j))
        removed_extra_list.append(new_list)
          
    key_list = [tuple(k) for k in removed_extra_list]
    ligand_dict = {}
    for key in key_list:
        if key not in ligand_dict:
            ligand_dict[key] = 1
        else:
            ligand_dict[key] += 1

    return {lig: count for lig, count in sorted(ligand_dict.items(), key=lambda item: -item[1])}
```

```python
make_ligand_count_dict('metalpdb_report.tsv')
```
