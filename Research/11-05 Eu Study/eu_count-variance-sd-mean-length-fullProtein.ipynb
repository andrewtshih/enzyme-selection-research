```python
import pandas as pd
import numpy as np
import statistics
import os

from Bio import SeqIO
```

Collect names of all fasta files downloaded for Eu. Put them into a list.

```python
filenames = []

for filename in os.listdir(os.getcwd() + '/eu_fasta_files'):
    with open(os.path.join(os.getcwd() + '/eu_fasta_files', filename)) as f:
        filenames.append(filename)

filenames
```

Collect names of all site IDs.

```python
code_list = ['1jhw_A', '1jhw_2', '1qsl_A', '2fby_1', '1wc4_A', '1wc4_2', '2fby_A', '1qsl_1', '2p1x_A', '4lp1_1', 
            '4lp1_3', '4lp1_6', '2pc2_A', '2pe7_A', '2yat_A', '3lgr_A', '3mvo_A', '3ojo_A', '3py9_A',
            '3tf6_C', '3tor_B', '3tor_C', '3tor_11', '3tor_18', '3tor_9', '4bad_A', '4bap_A', '4bar_A', '4lp1_B',
             '4lp1_A', '4lp1_6', '4lp1_10']
```

```python
# Collect information about each sequence responding to the sites in the code_list above.
rec_dict = {}
for filename in filenames:
    rec_dict.update(SeqIO.to_dict(SeqIO.parse("eu_fasta_files/" + filename, "fasta")))

# To prepare making the dataframe, make a list of lists (inner list contains site and the sequence)
seq_info = []
for seq_rec in rec_dict.values():
    if seq_rec.id in code_list:
        seq_info.append([seq_rec.id, seq_rec.seq])

# Remove '-' characters and convert sequences to type string
for i in seq_info:
    i[1] = str(i[1].replace('-', ''))
    
seq_info
```

```python
test_df = pd.DataFrame(seq_info, columns = ['Entity ID', 'Sequence'])
test_df
```

```python
rowNum = test_df.shape[0]
colNum = test_df.shape[1]
```

```python
out_meanvarstdlen = pd.DataFrame(columns=['Entity_ID','mean','var','std','len'], index=range(rowNum))
out_meanvarstdlen
```

```python
# fill the IDs
for i in range(rowNum):
    out_meanvarstdlen.iloc[i][0] = test_df.iloc[i][0]

out_meanvarstdlen
```

```python
for i in range(0, rowNum):                     # loop through all rows
    j = 1                                      # only one seq
    if isinstance(test_df.iloc[i,j],str):      # whether the cell is a string or NaN
        l = len(test_df.iloc[i,j])             # determine the length at the cell
        lst = [None] * l                       # create an empty list with the deterined length to be filled                     
        for p in range(l):                     # assign hydrophobicity number according to the sigmaaldrich website
            if test_df.iloc[i,j][p] == 'F':    # https://www.sigmaaldrich.com/US/en/technical-documents/technical-article/protein-biology/protein-structural-analysis/amino-acid-reference-chart#hydro
                lst[p] = 100
            elif test_df.iloc[i,j][p] == 'I':
                lst[p] = 99
            elif test_df.iloc[i,j][p] == 'W':
                lst[p] = 97
            elif test_df.iloc[i,j][p] == 'L':
                lst[p] = 97
            elif test_df.iloc[i,j][p] == 'V':
                lst[p] = 76
            elif test_df.iloc[i,j][p] == 'M':
                lst[p] = 74
            elif test_df.iloc[i,j][p] == 'Y':
                lst[p] = 63
            elif test_df.iloc[i,j][p] == 'C':
                lst[p] = 49
            elif test_df.iloc[i,j][p] == 'A':
                lst[p] = 41
            elif test_df.iloc[i,j][p] == 'T':
                    lst[p] = 13
            elif test_df.iloc[i,j][p] == 'H':
                lst[p] = 8
            elif test_df.iloc[i,j][p] == 'G':
                lst[p] = 0
            elif test_df.iloc[i,j][p] == 'S':
                lst[p] = -5
            elif test_df.iloc[i,j][p] == 'Q':
                lst[p] = -10
            elif test_df.iloc[i,j][p] == 'R':
                lst[p] = -14
            elif test_df.iloc[i,j][p] == 'K':
                lst[p] = -23
            elif test_df.iloc[i,j][p] == 'N':
                lst[p] = -28
            elif test_df.iloc[i,j][p] == 'E':
                lst[p] = -31
            elif test_df.iloc[i,j][p] == 'P':
                lst[p] = -46
            elif test_df.iloc[i,j][p] == 'D':
                lst[p] = -55
            else:                                # print else just in case
                difLetter = test_df.iloc[i,j][p]
                print('There is a', difLetter, 'at row', i, 'location', p)
        mean = statistics.mean(lst)              # calculate mean, std, var
        std = statistics.pstdev(lst)
        var = statistics.pvariance(lst)
        out_meanvarstdlen.iloc[i][1] = mean
        out_meanvarstdlen.iloc[i][2] = var
        out_meanvarstdlen.iloc[i][3] = std
        out_meanvarstdlen.iloc[i][4] = l
```
